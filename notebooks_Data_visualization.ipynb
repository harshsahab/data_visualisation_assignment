```python
#Generate a synthetic dataset with correlated features. Visualize the correlation matrix of a dataset using a heatmap.




import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

x = np.random.rand(100)
y = 2 * x + np.random.normal(0, 0.1, 100)
z = -x + np.random.normal(0, 0.1, 100)

data = np.column_stack((x, y, z))

correlation_matrix = np.corrcoef(data, rowvar=False)

plt.imshow(correlation_matrix, cmap='coolwarm', vmin=-1, vmax=1)

plt.colorbar(label='Correlation')
plt.xticks([0, 1, 2], ['Feature X', 'Feature Y', 'Feature Z'])
plt.yticks([0, 1, 2], ['Feature X', 'Feature Y', 'Feature Z'])

for i in range(3):
    for j in range(3):
        plt.text(j, i, f'{correlation_matrix[i, j]:.2f}',
                 ha='center', va='center')

plt.title('Correlation Matrix Heatmap')
plt.show()

```

```python
#Using the sales data, generate a heatmap to visualize the variation in sales across different months and days. np.random.seed(20)data = {'Month': np.random.choice(['Jan', 'Feb', 'Mar', 'Apr', 'May'], 100),'Day': np.random.choice(range(1, 31), 100),'Sales': np.random.randint(1000, 5000, 100)}df = pd.DataFrame(data)


import numpy as np
import matplotlib.pyplot as plt

np.random.seed(20)
data = {'Month': np.random.choice(['Jan', 'Feb', 'Mar', 'Apr', 'May'], 100),
        'Day': np.random.choice(range(1, 31), 100),
        'Sales': np.random.randint(1000, 5000, 100)}

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
heatmap = np.full((5, 30), np.nan)

for i, month in enumerate(months):
    for day in range(1, 31):
        sales = [data['Sales'][j] for j in range(100)
                 if data['Month'][j] == month and data['Day'][j] == day]
        if sales:
            heatmap[i, day - 1] = np.mean(sales)

plt.imshow(heatmap, cmap='YlOrRd', aspect='auto')
plt.colorbar(label='Sales')
plt.xticks(range(30), range(1, 31))
plt.yticks(range(5), months)
plt.xlabel('Day')
plt.ylabel('Month')
plt.title('Sales Heatmap')
plt.show()
```

```python
#.Create a Bokeh scatter plot using randomly generated x and y values. Use different sizes and colors for the markers based on the 'sizes' and 'colors' columns.

import numpy as np
from bokeh.plotting import figure, show
from bokeh.io import output_notebook

output_notebook()

# Generate random x and y values
np.random.seed(42)
x = np.random.randint(1, 100, 50)
y = np.random.randint(1, 100, 50)

# Generate sizes and colors
sizes = np.random.randint(10, 30, 50)
colors = np.random.choice(
    ['red', 'blue', 'green', 'orange', 'purple'], 50
)

# Create Bokeh scatter plot
p = figure(
    title="Bokeh Scatter Plot",
    x_axis_label="X Values",
    y_axis_label="Y Values",
    width=700,
    height=450
)

p.scatter(
    x=x,
    y=y,
    size=sizes,
    color=colors,
    alpha=0.7
)

show(p)
```

```python
#Create a Bokeh histogram to visualize the distribution of the given data. data_hist = np.random.randn(1000)hist, edges = np.histogram(data_hist, bins=30)

import numpy as np
from bokeh.plotting import figure, show
from bokeh.io import output_notebook

# Display plot in Jupyter Notebook
output_notebook()

# Generate random data
data_hist = np.random.randn(1000)

# Calculate histogram
hist, edges = np.histogram(data_hist, bins=30)

# Create Bokeh figure
p = figure(
    title="Histogram of Random Data",
    x_axis_label="Value",
    y_axis_label="Frequency",
    width=700,
    height=400
)

# Draw histogram bars
p.quad(
    top=hist,
    bottom=0,
    left=edges[:-1],
    right=edges[1:],
    fill_color="skyblue",
    line_color="white"
)

# Display histogram
show(p)
```

```python
#Create a Bokeh heatmap using the provided dataset. data_heatmap = np.random.rand(10, 10)x = np.linspace(0, 1, 10)y = np.linspace(0, 1, 10)xx, yy = np.meshgrid(x, y)

import numpy as np
from bokeh.plotting import figure, show
from bokeh.io import output_notebook
from bokeh.models import LinearColorMapper, ColorBar, ColumnDataSource
from bokeh.transform import transform
from bokeh.palettes import Viridis256

output_notebook()

# Generate dataset
data_heatmap = np.random.rand(10, 10)
x = np.linspace(0, 1, 10)
y = np.linspace(0, 1, 10)
xx, yy = np.meshgrid(x, y)

# Flatten arrays
x_values = xx.flatten()
y_values = yy.flatten()
values = data_heatmap.flatten()

# Create data source
source = ColumnDataSource(data=dict(
    x=x_values,
    y=y_values,
    value=values
))

# Map values to colors
color_mapper = LinearColorMapper(
    palette=Viridis256,
    low=0,
    high=1
)

# Create heatmap
p = figure(
    title="Bokeh Heatmap",
    x_axis_label="X",
    y_axis_label="Y",
    width=600,
    height=500
)

p.rect(
    x='x',
    y='y',
    width=0.1,
    height=0.1,
    fill_color=transform('value', color_mapper),
    line_color=None,
    source=source
)

# Add color bar
color_bar = ColorBar(
    color_mapper=color_mapper,
    title="Value"
)
p.add_layout(color_bar, 'right')

show(p)
```
